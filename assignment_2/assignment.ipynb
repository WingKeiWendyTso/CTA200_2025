```python
#part1
def f(x):
    """This function is a mathematical function that takes the input x and 
    return x^3-x^2-1.
    parameter: 
    x: float, input value"""
    return x**3-x**2-1

def df(x):
    """This function is a mathematical function that takes the input x and 
    return the derivative of function f(x), which is 3x^2-2x
    parameter: 
    x: float, input value"""
    return 3*x**2-2*x

print(f(3))
print(df(3))

#part2
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """This function is to use newton iterations to find the roots of f
    with its derivative df and return the number of iterations it
    takes for finding its root. 
    parameters:
    f: function
    df: derivative of the function
    x0: float, initial guess of the root
    epsilon: float, convergence tolerance
    max_iter: int, Maximum number of iterations
    
    return root of the function if successful, otherwise None.
    """
    x = x0
    for i in range(max_iter):    
        x=x - f(x)/df(x)
        if abs(f(x)) < epsilon :
            print(f"For x0 = {x0} with epsilon = {epsilon}")
            print(f"Found root in {i+1} iteration")
            print("value of x is ", x)
            return x
    print("Iteration failed")
    return None
        
#part3
print(newton(f, df,5))
print(newton(f, df,10))
print(newton(f, df,15))
print(newton(f, df,29))
print(newton(f, df,5, epsilon=1e-8))
print(newton(f, df,10, epsilon=1e-8))
print(newton(f, df,15, epsilon=1e-8))
print(newton(f, df,29, epsilon=1e-8))
```
